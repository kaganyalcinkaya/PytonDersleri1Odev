```python
not1 = float(input("1. ders notunu girin: "))
not2 = float(input("2. ders notunu girin: "))
not3 = float(input("3. ders notunu girin: "))
not4 = float(input("4. ders notunu girin: "))

ortalama = (not1 + not2 + not3 + not4) / 4

print("Ortalama:", ortalama)

if ortalama >= 90:
    harf = "AA"
elif ortalama >= 85:
    harf = "BA"
elif ortalama >= 80:
    harf = "BB"
elif ortalama >= 70:
    harf = "CB"
elif ortalama >= 60:
    harf = "CC"
elif ortalama >= 50:
    harf = "DC"
elif ortalama >= 40:
    harf = "DD"
else:
    harf = "FF"

print("Harf Notu:", harf)

en_yuksek = max(not1, not2, not3, not4)
en_dusuk = min(not1, not2, not3, not4)

print("En yüksek not:", en_yuksek)
print("En düşük not:", en_dusuk)
```

```python
sepet = [80, 150, 250, 120, 90, 300]


yuz_tl_uzeri = []

for fiyat in sepet:
    if fiyat > 100:
        yuz_tl_uzeri.append(fiyat)

print("100 TL üzerindeki ürünler:", yuz_tl_uzeri)


# Toplam sepet tutarını hesaplama
toplam = 0

for fiyat in sepet:
    toplam += fiyat

print("Sepet toplamı:", toplam, "TL")


# 500 TL üzeri ise %10 indirim
if toplam > 500:
    print("%10 indirim kazandınız!")

    indirim = toplam * 0.10
    yeni_tutar = toplam - indirim

    print("İndirim tutarı:", indirim, "TL")
    print("Yeni sepet tutarı:", yeni_tutar, "TL")
else:
    print("İndirim uygulanmadı.")
```
