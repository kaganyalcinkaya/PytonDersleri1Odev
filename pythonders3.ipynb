```python
class Araba:
    def __init__(self, marka, model, hiz):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis

    def bilgi_goster(self):
        print("markası:", self.marka)
        print("modeli:", self.model)
        print("hızı:", self.hiz, "km/s")



```

```python
araba1 = Araba("lamborghini", "huracan", 250)
araba1.bilgi_goster()
araba1.hizlandir(20)

print("Hızı artırıldıktan sonra :")
araba1.bilgi_goster()
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.uzunluk = uzunluk
        self.genislik = genislik

    def alan_hesapla(self):
        alan = self.uzunluk * self.genislik
        return alan

    def cevre_hesapla(self):
        cevresi = 2 * (self.genislik + self.uzunluk)
        return 2 * (self.uzunluk + self.genislik)


dikdortgen1 = Dikdortgen(15, 4)
print("Alan:", dikdortgen1.alan_hesapla())
print("Çevre:", dikdortgen1.cevre_hesapla())
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"

kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)

print(kitap1.ozet())
```
